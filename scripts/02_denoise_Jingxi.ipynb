`processed/02_denoise/`

- denoising statistics summary (`dada2_stats.qzv`)

```python
import pandas as pd
from qiime2 import Visualization, Artifact
from qiime2 import Visualization

raw_dir = '../data/raw'
in_dir  = '../data/processed_ignored/01_preprocessing'
out_dir = '../data/processed_ignored/02_denoise'
vis_dir = '../data/processed/02_denoise'
```

```python
!mkdir -p $out_dir $vis_dir
```

```python
!qiime dada2 denoise-single \
  --i-demultiplexed-seqs $in_dir/reads_trimmed.qza \
  --p-trunc-len 0 \
  --p-n-threads 3 \
  --o-table $out_dir/dada2_table.qza \
  --o-representative-sequences $out_dir/dada2_rep_set.qza \
  --o-denoising-stats $out_dir/dada2_stats.qza \
  --o-base-transition-stats $out_dir/dada2_base_transition_stats.qza
```

```python
!qiime metadata tabulate \
  --m-input-file $out_dir/dada2_stats.qza \
  --o-visualization $vis_dir/dada2_stats.qzv
```

```python
Visualization.load(f'{vis_dir}/dada2_stats.qzv')
```

```python
!qiime feature-table tabulate-seqs \
  --i-data $out_dir/dada2_rep_set.qza \
  --o-visualization $vis_dir/dada2_rep_set.qzv
```

```python
Visualization.load(f'{vis_dir}/dada2_rep_set.qzv')
```

```python
!qiime feature-table summarize \
  --i-table $out_dir/dada2_table.qza \
  --m-metadata-file $raw_dir/fungut_metadata.tsv \
  --o-feature-frequencies $out_dir/dada2_table_feature_frequencies.qza \
  --o-sample-frequencies $out_dir/dada2_table_sample_frequencies.qza \
  --o-summary $vis_dir/dada2_table.qzv
```

```python
Visualization.load(f'{vis_dir}/dada2_table.qzv')
```
