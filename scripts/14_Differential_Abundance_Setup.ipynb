# 14 Differential Abundance Setup

We prepare an unrarefied fungal genus count table for the differential abundance analyses. We use the fungal ASV table and taxonomy from notebook 03 and sum ASV counts within each genus for each sample.

The age, settlement, diet, and other differential abundance notebooks use this shared table. Each analysis notebook selects the samples needed for its own models.

## 1. Setup

```python
import pandas as pd
import qiime2 as q2
from qiime2 import Artifact
```

```python
in_dir = "../data/processed_ignored/03_taxonomy"
out_dir = "../data/processed_ignored/14_differential_abundance_setup"

fungal_table = f"{in_dir}/table_fungi_filtered.qza"
taxonomy = f"{in_dir}/taxonomy.qza"

genus_table = f"{out_dir}/table_fungi_l6.qza"

!mkdir -p "$out_dir"
```

## 2. Prepare the fungal genus count table

### 2.1 Collapse ASV counts to genus

```python
!qiime taxa collapse \
    --i-table "$fungal_table" \
    --i-taxonomy "$taxonomy" \
    --p-level 6 \
    --o-collapsed-table "$genus_table"
```

```python
collapsed_genus = Artifact.load(genus_table).view(pd.DataFrame)
collapsed_genus
```
