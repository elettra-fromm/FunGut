```python
print("CRS:", ghsl.crs)
print("Resolution:", ghsl.res)
print("Bounds:", ghsl.bounds)
print("Dimensions:", ghsl.width, ghsl.height)
print("NoData value:", ghsl.nodata)
```

```python
#Transform sample coordinates into matching coordination system
```

```python
samples_ghsl = samples.to_crs(ghsl.crs)
```

```python
# Get coordinates of each sample in the GHSL coordinate system
coords = [(point.x, point.y) for point in samples_ghsl.geometry]

# Extract raster values
values = [val[0] for val in ghsl.sample(coords)]

# Add values to a copy of your metadata
df_ghsl = samples_ghsl.copy()
df_ghsl['GHSL_code'] = values
```

```python
df_ghsl['GHSL_code'].value_counts().sort_index()
```

```python
#Classify into urban, suburban, rural from 7 level classification:
```

```python
urban_codes = [30, 23]
suburban_codes = [22, 21]
rural_codes = [13, 12, 11]

df_ghsl['Settlement_Type'] = 'Unclassified'

df_ghsl.loc[df_ghsl['GHSL_code'].isin(urban_codes), 'Settlement_Type'] = 'Urban'

df_ghsl.loc[df_ghsl['GHSL_code'].isin(suburban_codes), 'Settlement_Type'] = 'Suburban'

df_ghsl.loc[df_ghsl['GHSL_code'].isin(rural_codes), 'Settlement_Type'] = 'Rural'
```

```python
#Inspect neighboring 1kmx1km cells of the 3 unclassified samples:
```

```python
import numpy as np
import pandas as pd

def inspect_neighbours(row, raster, radius=4):

    # Transform sample coordinates to the raster CRS
    point = gpd.GeoSeries(
        [gpd.points_from_xy([row['longitude_sample']], [row['latitude_sample']])[0]],
        crs='EPSG:4326'
    ).to_crs(raster.crs).iloc[0]

    # Find the raster pixel containing the sample
    raster_row, raster_col = raster.index(point.x, point.y)

    # Read surrounding pixels (3 x 3 if radius=1)
    window = raster.read(
        1,
        window=(
            (raster_row-radius, raster_row+radius+1),
            (raster_col-radius, raster_col+radius+1)
        )
    )

    return window
```

```python
for index, sample in water_samples.iterrows():

    print(f"\nSample: {sample['ID']}")
    print(f"Country: {sample['country_sample']}")

    neighbours = inspect_neighbours(sample, ghsl)

    print(neighbours)
```

```python
def classify_neighbourhood(matrix):

    urban = np.isin(matrix, [30, 23]).sum()
    suburban = np.isin(matrix, [22, 21]).sum()
    rural = np.isin(matrix, [13, 12, 11]).sum()
    water = np.isin(matrix, [10]).sum()

    counts = {
        'Urban': urban,
        'Suburban': suburban,
        'Rural': rural,
    }

    # Find the category with the most pixels
    return max(counts, key=counts.get)
```

```python
for index, sample in water_samples.iterrows():

    matrix = inspect_neighbours(sample, ghsl, radius=5)

    classification = classify_neighbourhood(matrix)

    print(sample['ID'], classification)
```
