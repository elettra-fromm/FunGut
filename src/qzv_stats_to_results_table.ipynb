```python
# Unadjusted, 16mers

from zipfile import ZipFile

from statsmodels.stats.multitest import multipletests

pair_labels = ["Female vs. Male", "Female vs. Other", "Male vs. Other"]
pair_results = []

for pair_number, label in enumerate(pair_labels, start=1):
    qzv_path = f"{unadjusted_permanova_dir}/kmer16_pair_{pair_number:02d}.qzv"

    with ZipFile(qzv_path) as archive:
        tsv = next(
            name for name in archive.namelist()
            if name.endswith("/data/adonis.tsv")
        )
        with archive.open(tsv) as file:
            results = pd.read_csv(file, sep="\t", index_col=0)

    sex = results.loc["sex_sample"]
    n = int(results.loc["Total", "Df"]) + 1
    pair_results.append((label, n, sex["R2"], sex["F"], sex["Pr(>F)"]))

raw_p = [result[4] for result in pair_results]
holm_p = multipletests(raw_p, method="holm")[1]

print("#### Unadjusted pairwise results · 16-mers\n")
print("| Sex pair | n | R² | Pseudo-F | Raw p | Holm-adjusted p |")
print("| --- | ---: | ---: | ---: | ---: | ---: |")

for (label, n, r2, pseudo_f, p), adjusted_p in zip(pair_results, holm_p):
    holm_text = f"{adjusted_p:.3f}"
    if adjusted_p < 0.05:
        holm_text = f"**{holm_text}**"

    print(f"| {label} | {n} | {r2:.4f} | {pseudo_f:.3f} | {p:.3f} | {holm_text} |")
```
